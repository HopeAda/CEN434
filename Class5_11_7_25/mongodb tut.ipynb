```python
import pymongo


myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydb"]
mydbs = (myclient.list_database_names())
```

```python
if "mydb" in mydbs:
    print("Exists")
```

```python
doc = mydb["students"]
data = {"name":"Jane Stan", "age":32, "course":"Mechanical Engineering"} 
y = doc.insert_one(data)
print(y)
```

```python
myq = {"name" : {"J"}}
myans = doc.find(myq)
for n in myans:
    print(n)
```
